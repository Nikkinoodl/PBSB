## Convert New Blog Post Content in SQLLite
Blog posts for the new website are created using Blogifier. We then take the contents of the Blogifier database and convert it to a flat file, discard unwanted fields and change some parts of the image tags. We then transform these posts and dynamically generate new website content using 11ty.

The content largely follows that of the process used to convert the old blog posts from the nopCommerce/PBSB/Blogifier website, although some of the fields and field names are different.


### Imports

```python
import sqlite3
import csv
import pandas as pd
import os
import json
import shutil
import filecmp
from pathlib import Path
```

### Read Blog Posts Data from Database
First, read the blog posts from the PBSB database and write into a flat CSV output file.

```python
# Connect to the SQLite database
conn = sqlite3.connect('C:\\Users\\simon\\Projects\\Blogifier\\src\\Blogifier\\Blog.db')
cursor = conn.cursor()

# Query the data
#cursor.execute("SELECT * FROM BlogPosts")
cursor.execute("""SELECT 
    b.*,
    GROUP_CONCAT(c.content, ',') AS categories_yaml
    FROM Posts b
    LEFT JOIN PostCategories pc ON b.Id = pc.PostId
    LEFT JOIN Categories c ON pc.CategoryId = c.Id
    GROUP BY b.Id;""")

fileName = 'newblogposts.csv'

# Write to CSV
with open(fileName, 'w', newline='') as csv_file:
    writer = csv.writer(csv_file, quoting=csv.QUOTE_ALL, lineterminator='\n')
    # For this conversion, we skip the headers (can be useful during development)
    # writer.writerow([i[0] for i in cursor.description])
    # Write data
    writer.writerows(cursor.fetchall())

conn.close()    
```

### Process the CSV File
Now switch to Pandas for easier processing of the data.

```python
# Read the CSV file making sure we omit the header row
df = pd.read_csv(fileName, encoding='windows-1252', header=None)

# Replace NaN values in column 16 with "None"
df[16] = df[16].fillna("None")

df.head()
```

Delete unnecessary fields. Data does not need to be reorganized.

```python
df = df.drop(df.columns[[0, 1, 2, 8, 9, 10, 11, 13, 14, 15]], axis=1)
#df.reindex()
df.head()
```

### Add Column Names and Sort by Date

```python
df.columns = ['Title', 'Slug', 'Description', 'Content', 'Image', 'Date', 'Categories']

# Sort by 'Age' in descending order
df = df.sort_values(by='Date', ascending=False)

df.head()
```

### Wrangle the Data

Convert date-time data to required format YYYY-MM-DD HH:MM:SS

```python
# Convert column at index 5 to datetime
col_name = df.columns[5]
df[col_name] = pd.to_datetime(df[col_name], format = 'ISO8601')

#format='%d/%m/%y %H:%M:%S')

# Format to 'YYYY-MM-DD HH:MM:SS'
df[col_name] = df[col_name].dt.strftime('%Y-%m-%d %H:%M:%S')
```

### Clean and Validate the Data

Change the image folder name to match PBSB website usage.

```python
col_name = df.columns[3]
df[col_name]=df[col_name].str.replace('/data', '/images', regex=False)
df[col_name]=df[col_name].str.replace('/img', '/images', regex=False)

col_name = df.columns[4]
df[col_name]=df[col_name].str.replace('/data', '/images', regex=False)
df[col_name]=df[col_name].str.replace('/img', '/images', regex=False)
```

### Final Visual Inspection of the Data
Last chance to visually inspect the data before handing it over to 11ty.

```python
df.head()
```

### Write Out the Data to a Unix-type CSV file (has LF line endings, not CRLF)

Do this so we have an easily readable copy of the output on hand for reference. It is otherwise not used for the website.

```python
# Write the cleaned-up data to the CSV file. Make sure there is a header row!

df.to_csv(fileName, index=False, header=True, lineterminator='\n')
```

## Write Out Individual Blog Post Files

Create the individual files that will contain each blog post.

```python
# Output directory
output_dir = "C:\\Users\\simon\\Projects\\PBSB\\blog"
os.makedirs(output_dir, exist_ok=True)
```

Data has already been cleaned and formatted so it can be written directly to an output file.

```python
# Write each row to its own text file
for idx, row in df.iterrows():

    cats = str(row['Categories']).split(',')
    file_path = os.path.join(output_dir, str(row['Slug']) + ".liquid")

    try:
        with open(file_path, "w", encoding="utf-8") as f:
            f.write("--- \n")
            f.write("layout: post.njk" + "\n")
            f.write("title: \"" + str(row['Title']).replace('"', '\\"') + "\"\n")
            f.write("permalink: /blog/" + str(row['Slug']) + "/ \n")
            f.write("date: " + str(row['Date']) + "\n")
            f.write("shortdate: " + str(pd.to_datetime(row['Date']).strftime("%d %b %Y")) + "\n")
            f.write("image: " + str(row['Image']) + "\n")
            f.write("description: \"" + str(row['Description']).replace('"', '\\"') + "\"\n")
            f.write("categories:\n")
            for c in cats:
                f.write(f"  - {c.strip()}\n")
            f.write("tags: post" + "\n")
            f.write("--- \n")
            f.write("\n")
            f.write("\n")
            f.write(str(row['Content']))
    except OSError as e:
        print(f"Error writing file {file_path}: {e}")

print(f"{len(df)} files written to '{output_dir}'")
```

## Final Steps - Copy Image Files and Attachments

Copy files and folders from the Blogifier solution over to the PBSB images folder if they don't already exist.

```python
def copy_if_missing(src, dst):
    src = Path(src)
    dst = Path(dst)

    if not src.is_dir():
        raise ValueError("Source must be a directory")

    # Compare directory trees
    comparison = filecmp.dircmp(src, dst)

    # Files only in source → copy them
    for file in comparison.left_only:
        src_file = src / file
        dst_file = dst / file
        if src_file.is_file():
            shutil.copy2(src_file, dst_file)
            print(f"Copied new file: {src_file} -> {dst_file}")
        elif src_file.is_dir():
            shutil.copytree(src_file, dst_file)
            print(f"Copied new folder: {src_file} -> {dst_file}")

    # Recurse into common subdirectories
    for common_dir in comparison.common_dirs:
        copy_if_missing(src / common_dir, dst / common_dir)

# Copy files across:
copy_if_missing(r"C:\Users\Simon\Projects\Blogifier\src\Blogifier\wwwroot\data", r"C:\Users\Simon\Projects\PBSB\images")

```
